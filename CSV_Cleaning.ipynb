```python
import pandas as pd
import numpy as np
import re
from google.colab import files
```

```python
uploaded = files.upload()
file_name = next(iter(uploaded))
```

```python
df = pd.read_csv(file_name)
```

```python
df = df.drop_duplicates()

# Fill missing values
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
df['rating'] = df['rating'].fillna('Unrated')
df['duration'] = df['duration'].fillna('Unknown')
df['listed_in'] = df['listed_in'].fillna('Unknown')
df['description'] = df['description'].fillna('No Description')
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

# Fill missing dates
df['date_added'] = df['date_added'].fillna(pd.Timestamp('2020-01-01'))
```

```python
df['Genre Clean'] = df['listed_in'].apply(
    lambda x: str(x).split(',')[0].strip()
)
```

```python
df['Country Clean'] = df['country'].apply(
    lambda x: str(x).split(',')[0].strip()
)
```

```python
def extract_minutes(duration):
    if pd.isna(duration):
        return np.nan

    match = re.search(r'(\d+)', str(duration))

    if match:
        return int(match.group(1))

    return np.nan

df['Runtime Number'] = df['duration'].apply(extract_minutes)
```

```python
df['Title Count'] = 1
```

```python
df = df.rename(columns={
    'show_id': 'Show ID',
    'type': 'Content Type',
    'title': 'Title',
    'director': 'Director',
    'cast': 'Cast',
    'country': 'Country',
    'date_added': 'Date Added',
    'release_year': 'Release Year',
    'rating': 'Rating',
    'duration': 'Duration',
    'listed_in': 'Genre',
    'description': 'Description'
})
```

```python
print("DATA CLEANING COMPLETE")
print()
print("Dataset Shape:", df.shape)
print()
print(df.head())
```

```python
output_file = 'NETFLIX_TABLEAU_CLEANED.csv'

df.to_csv(output_file, index=False)

print()
print(f'Exported File: {output_file}')
```
